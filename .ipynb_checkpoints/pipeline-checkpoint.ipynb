```python
import pandas as pd

metadata = pd.read_csv('data/metadata.csv', index_col=0)
metadata
```

```python
protein_data = pd.read_csv('data/proteomics.txt', header=0, index_col=0, sep='\t').T
protein_data.index_name = 'Patient_ID'
protein_data
```

```python
combined_df = metadata.merge(protein_data, 
               left_index=True,
               right_index=True,
               how='inner')
combined_df
```
